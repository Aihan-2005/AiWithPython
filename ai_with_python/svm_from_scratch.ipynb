```python
import numpy as np
from typing import Tuple, Optional
import warnings

class SVMFromScratch:
    
    def __init__(self,learning_rate:float=0.001,n_iterations:int=1000,C:float=1.0,batch_size:Optional[int]=None,verbose:bool=False):
        
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations 
        self.C = C
        self.batch_size = batch_size
        self.random_state = self.random_state
        self.verbose = verbose
        
        
        self.w_ = None
        self.b_ = None
        
        
        self.support_vectors = None
        self.support_vectors_indices_ = None
        
        self.loss_history_ = []
        
    def _initialize_parameters(self,n_features:int)-> None:
        
        
        if self.random_state is not None:
            np.random.seed(self.random_state)
            
            
        
        self.w_ = np.random.randn(n_features)*0.01
        self.b_ = 0.0
        
    def _compute_hinge_loss(self,X:np.ndarray,y:np.ndarray)->float:
        
        
        n_samples = X.shape[0]
        
        predictions = np.dot(X,self.w_) + self.b_
        
        margins = y*predictions
        
        hinge_loss = np.maximum(0,1-margins)
        
        
        regularization_term = 0.5*np.dot(self.w_,self.w_)
        empirical_loss = self.C *np.sum(hinge_loss)
        
        
        total_loss =regularization_term+empirical_loss
        return total_loss
    
    
    
    
    def _compute_gradients(
        self,X_batch:np.ndarray,y_batch:np.ndarray
    )->Tuple[np.ndarray,float]:
        
        
        n_samples = X_batch.shape[0]
        
        predictions = np.dot(X_batch,self.w_) + self.b_
        margins = y_batch*predictions
        
        
        margin_violations = margins<1
        
        
        grad_w = self.w_.copy()
        
        
        if np.any(margin_violations):
            grad_w -= self.C*np.dot(
                y_batch[margin_violations],
                X_batch[margin_violations]
            )
            
            
        grad_b =0.0
        if np.any(margin_violations):
            grad_b = -self.C*np.sum(y_batch[margin_violations])
            
        return grad_w,grad_w
    
    
    
    def fit(self,X:np.ndarray,y:np.ndarray):
        
        
        n_samples = X.shape[0]
        
        if not self.is_fitted_:
            self._initialize_parameters(X.shape[1])
            
            
        indices = np.arange(n_samples)
        np.random.shuffle(indices)
        
        for idx in indices:
            self.partial_fit(X[idx:idx+1],y[idx:idx+1])
            
        return self
    
    
    
    def decision_function(self,X:np.ndarray)-> np.ndarray:
        
        
        if not self.is_fitted_:
            raise RuntimeError('model most be fitted before prediction')
        
        return np.dot(X,self.w_) + self.b_
    
    
    def predict(self,X:np.ndarray)-> np.ndarray:
        
        
        descition = self.decision_function(X)
        predictions  = np.sign(descition)
        predictions[predictions == 0] = 1
        return predictions
    
    def score(self,X:np.ndarray,y:np.ndarray)->float:
        
        predictions = self.predict(X)
        return np.mean(predictions == y)
    
    def generate_streaming_data(n_samples:int=500,random_state:int=6):
    
    
        X,y_binary = make_blobs(
            n_samples=n_samples,
            n_features=2,
            centers=2,
            cluster_std = 1.2,
            center_box=(-5,5),
            random_state=random_state
        )
    
    
        y = 2*y_binary-1
    
    
        scaler = StandardScaler()
        X = scaler.fit_transform(X)
        
        return X,y
    
def simulate_online_learning(X,y,batch_size:int=1):
    
    n_samples = X.shape[0]
    
    custom_svm = CustomOnlineSVM(
        learning_rate =0.01,
        C=1.0,
        n_iterations= 1,
        random_state=6,
        verbose = False
    )
    
    
    sklearn_svm = SGDClassifier(
        loss='hing',
        penalty='l2',
        alpha ='0.001',
        learning_rate='optimal',
        max_iter =1,
        tol = None,
        random_state=6,
        warm_start = True
    )
    
```
