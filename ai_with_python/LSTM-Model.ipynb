```python
import nltk 
nltk.download('gutenberg')
from nltk.corpus import gutenberg
```

```python
data = gutenberg.raw("shakespeare-hamlet.txt")

with open("datasets/hamlet.txt", "w") as file:
    file.write(data)
```

```python
import numpy as np 
from tensorflow.keras.preprocessing.text import Tokenizer 
from tensorflow.keras.preprocessing.sequence import pad_sequences 
from sklearn.model_selection import train_test_split 
```

```python
with open('datasets/hamlet.txt','r') as file:
    text = file.read().lower()
    
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index) + 1
total_words
```

```python
tokenizer.word_index
```

```python
input_sequences = []

for line in text.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1,len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```

```python
max_sequence_lenght = max([len(x) for x in input_sequences])
max_sequence_lenght
```

```python
input_sequences = np.array(pad_sequences(input_sequences,max_sequence_lenght))
input_sequences
```

```python
import tensorflow as tf
X,y = input_sequences[:,:-1],input_sequences[:,-1]
```

```python
X
```

```python
y
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_calsses = total_words)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
print(X.shape)
print(X_train.shape)
print(X_test.shape)
```

```python
print(y.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from tenserflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense,Dropout

model = Sequential()
model.add(Embedding(total_words,100,input_lenght=max_sequence_lenght-1))
model.add(LSTM(150,return_sequence=True))
model.add(Dropout(0.2))
model.add(LSTM(100))
model.add(Dense(total_words,activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model.summery()
```

```python
from tenserflow.kerase.callbacks import EarlyStopping
early_stopping_callback = EarlyStopping(monitor='val_loss',patience=3,restore_best_weight=True)
```

```python
history = model.fit(X_train,y_train,epochs=100,validation_data=(X_test,y_test),verbose=1,callbacks=[early_stopping_callback])
```

```python
def predict_next_word(model,tokenizer,text,max_sequence_len):
    token_list = tokenizer.text_to_sequence([text])[0]
    if len(token_list) >= max_sequence_len:
        token_list = token_list[-(max_sequence_len-1)]
    
    token_list = pad_sequences([token_list],maxlen=max_sequence_len-1,padding='pre')
    predicted = model.predict(token_list,verbose=0)
    predict_next_word = np.argmax(predicted,axis=1)
    for word,index in tokenizer.word_index.items():
        if index == predict_next_word:
            return word
        return None
    
```

```python
sample_text = 'To be or not to be'
print(f'Sample text:{sample_text}')
max_sequence_lenght = model.input_shape[1] + 1
predict_next_word = predict_next_word(model,tokenizer,sample_text,max_sequence_lenght)
print(f'Next Word is :{predict_next_word}')
```
