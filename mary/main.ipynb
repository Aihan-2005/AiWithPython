```python
import osmnx as ox
```

```python
# Enable caching to avoid repeated API downloads
ox.settings.use_cache = True

# Center of our study area (latitude, longitude)
center = (51.5240, -0.1325)

# Download the driving network around this point
G = ox.graph.graph_from_point(
    center_point=center,
    dist=1200,
    network_type="drive",
    simplify=True
)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
print("Directed:", G.is_directed())
print("Multigraph:", G.is_multigraph())

```
