# Step 2 — Origin and Destination

```python
import sys
from IPython.display import display

print("Python executable:", sys.executable)

from routing.config import ORIGIN,ORIGINS,DESTINATION
from routing.map_loader import load_road_graph,nearest_drive_node,node_latlon
from routing.map_view import interactive_map,snap_distance_m
```

```python
G = load_road_graph()
print('GRAPH:',type(G).__name__)
print('NODES:',G.number_of_nodes())
print('DIRECT EDGES:', G.number_of_edges())
```

```python
for name,latlon in ORIGINS.items():
    node = nearest_drive_node(G,*latlon)
    print(f'{name}: node={node},snapped={node_latlon(G,node)}',
            f"offset={snap_distance_m(G,*latlon,node):.1f}m")


dest_node = nearest_drive_node(G,*DESTINATION)

print(f"Destination: node={dest_node}, snapped={node_latlon(G, dest_node)}, "
      f"offset={snap_distance_m(G, *DESTINATION, dest_node):.1f} m")

```

```python
from IPython.display import display
from ipywidgets import IntSlider
from ipyleaflet import Map

display(IntSlider(description="Widget Test"))

simple_map = Map(
    center=(51.4994, -0.1760),
    zoom=14
)

display(simple_map)
```

```python
widget,state = interactive_map(
    G,
    origin=ORIGIN,
    destination=DESTINATION,
    max_snap_distance_m=350.0
)

display(widget)
```

```python
print("Current origin node:", state["origin_node"])
print("Destination node:", state["destination_node"])
print("Origin snapped coordinates:", state["origin_latlon"])
print("Destination snapped coordinates:", state["destination_latlon"])
print("Last double-click (raw lat/lon):", state["last_click_latlon"])

assert state["origin_node"] in G
assert state["destination_node"] in G
```

```python
print("Origin:", state["origin_node"])
print("Destination:", state["destination_node"])
print("Last clicked:", state["last_click_latlon"])

assert state["origin_node"] in G
assert state["destination_node"] in G
assert state["last_click_latlon"] is not None

print("PASS: Origin & Destination")
```
