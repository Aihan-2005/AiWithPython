# Step 1 — OSMnx drivable network




```python
import osmnx as ox
from routing.map_loader import load_road_graph
from routing.config import GRAPH_PATH

G = load_road_graph()
print("OSMnx:", ox.__version__)
print("Graph type:", type(G).__name__)
print("Nodes:", G.number_of_nodes(), "Directed edges:", G.number_of_edges())
print("CRS:", G.graph["crs"])
print("Graph saved:", GRAPH_PATH)
```

```python
u, v, k, attrs = next(iter(G.edges(keys=True, data=True)))
print("Edge ID:", (u, v, k))
print("Length, m:", attrs["length"])
print("Highway:", attrs.get("highway"))
print("One-way:", attrs.get("oneway"))
```

```python
fig, ax = ox.plot_graph(G, node_size=0, edge_color="#64748b", edge_linewidth=1.3, bgcolor="white", show=False)
fig
```
